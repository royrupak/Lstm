```python
from pydub import AudioSegment
from pydub.utils import make_chunks
def split_audio(input_file, chunk_length_ms, output_dir):
  """
  Args:
    input_file: Path to the input audio file.
    chunk_length_ms: Length of each chunk in milliseconds.
    output_dir: Directory to save the output chunks.
  """
  audio = AudioSegment.from_file(input_file, format="m4a")
  chunks = make_chunks(audio, chunk_length_ms)

  for i, chunk in enumerate(chunks):
    chunk_name = f"Recording{i}.m4a"
    chunk_path = f"{output_dir}/{chunk_name}"
    chunk.export(chunk_path, format="wav")


input_audio_file = "/content/Recording.m4a"
chunk_duration_ms = 10000 # 10 seconds
output_directory = "/content/output/"

split_audio(input_audio_file, chunk_duration_ms, output_directory)
```

```python
import librosa
import numpy as np

for i in range(0,541):
  filename=f"Recording{i}.m4a"
  print("/content/output/"+filename)
  y, sr = librosa.load("/content/output/"+filename, sr=None)
  # 1. SNR estimate
  signal_energy = np.sum(y ** 2)
  noise_estimate = librosa.effects.split(y, top_db=30)
  noise_energy = np.sum([np.sum(y[start:end] ** 2) for start, end in noise_estimate])
  snr = 10 * np.log10(signal_energy / (noise_energy + 1e-6))
  # 2. Spectral flatness
  flatness = np.mean(librosa.feature.spectral_flatness(y=y))
  # 3. Zero-crossing rate
  zcr = np.mean(librosa.feature.zero_crossing_rate(y))

  print(f"SNR: {snr:.2f} dB, Flatness: {flatness:.3f}, Zero-Crossing Rate: {zcr:.3f}")

```

```python
#https://github.com/unwraps/simple-audio-spectrogram
#https://github.com/tharnob/spectrogram

import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np



for i in range(0,9):
  filename=f"Recording{i}.m4a"
  print("/content/output/"+filename)
  # Load audio
  y, sr = librosa.load("/content/output/"+filename, sr=None)

  # Display waveform
  plt.figure(figsize=(14, 5))
  librosa.display.waveshow(y, sr=sr)
  plt.title('Waveform of Recording'+str(i)+'.m4a')
  plt.xlabel('Time (s)')
  plt.ylabel('Amplitude')
  plt.show()

  # Display spectrogram
  plt.figure(figsize=(14, 5))
  D = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)
  librosa.display.specshow(D, y_axis='linear', x_axis='time', sr=sr)
  plt.colorbar(format='%+2.0f dB')
  plt.title('Spectrogram of Recording'+str(i)+'.m4a')
  plt.show()
```

##Pytorch Implementation of CleanUNet

This repo contains official PyTorch implementation of CleanUNet: Speech Denoising in the Waveform Domain with Self-Attention. CleanUNet is a causal speech denoising model on the raw waveform. It is based on an encoder-decoder architecture combined with several self-attention blocks to refine its bottleneck representations, which is crucial to obtain good results. The model is optimized through a set of losses defined over both waveform and multi-resolution spectrograms. The proposed method outperforms the state-of-the-art models in terms of denoised speech quality from various objective and subjective evaluation metrics. Sound demos can be found in This blog.

```python
import json5
import torch
from CleanUNet.network import CleanUNet

# Load config that matches the checkpoint
with open('configs/DNS-large-high.json', 'r') as f:
    config = json5.load(f)

network_config = config['network_config']

# Build the model exactly as trained
model = CleanUNet(**network_config)

# Load the pretrained checkpoint
ckpt = torch.load('exp/DNS-large-high/checkpoint/pretrained.pkl', map_location='cpu')
model.load_state_dict(ckpt['model_state_dict'])
model.eval()
```

```python
import torchaudio

# 2. Load and preprocess noisy audio
noisy, sr = torchaudio.load("/content/output/Recording0.m4a")   # shape [channels, time]
# Mono and resample to 32kHz if needed
if noisy.shape[0] > 1:
    noisy = noisy.mean(dim=0, keepdim=True)
if sr != 32000:
    resampler = torchaudio.transforms.Resample(sr, 32000)
    noisy = resampler(noisy)
# Add batch dimension: [1,1,T]
noisy = noisy.unsqueeze(0)

# 3. Run inference
with torch.no_grad():
    denoised = model(noisy)        # [1,1,T] output
denoised = denoised.squeeze(0)     # [1,T]
denoised = denoised.cpu()

# 4. Save output
# (Here we use torchaudio to write; any wav writer works)
torchaudio.save("/content/cleaned/Recording0.wav", denoised, 32000)
print("Denoised audio saved to 'denoised_output.wav'")

```
